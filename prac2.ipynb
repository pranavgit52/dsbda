```python
import numpy as np

# Sample data (can be a list, pandas Series, etc.)
data = [72, 69, 80, 65, 90, 76, 85, 100, 110, 95, 120, 130]

# Step 1: Calculate the mean and standard deviation
mean = np.mean(data)
std_dev = np.std(data)

# Step 2: Compute the Z-score for each data point
z_scores = [(x - mean) / std_dev for x in data]

# Step 3: Identify outliers (Z-score > 3 or Z-score < -3)
outliers = [data[i] for i in range(len(data)) if abs(z_scores[i]) > 3]

# Print results
print(f"Mean: {mean}")
print(f"Standard Deviation: {std_dev}")
print(f"Z-scores: {z_scores}")
print(f"Outliers: {outliers}")
```
